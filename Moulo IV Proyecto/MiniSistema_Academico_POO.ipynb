# PROYECTO MINI SISTEMA ACADEMICO CON POO

## Objetivo

Aplicar de forma integral los conceptos de clases, objetos, atributos, metodos, encapsulamiento y metodos especiales para crear un sistema simple de gestión de estudiantes, materias y calificaciones

1. ETRUCTURA DEL SISTEMA

| CLASE | ATRIBUTOS | MÉTODOS PRINCIPALES |
|-------|-----------|-------------------|
| Alumno | cedula, nombre, edad, nota | puede_votar(), obtener_calificacion() |
| Materia | nombre, creditos, docente | asignar_alumno(), listar_estudiantes() |
| Sistema | materias(lista) | crear_materia(), ver_materias() |

2. Requisitos funcionales

* Registrar alumnos y asignarlos a materias
* Listar los estudiantes de una materia
* Consultar si un alumno aprueba la materia
* Simular si puede votar por su edad
* Mostrar resumen por alumno y por materia

## ENUNCIADO

1. Define una clase ALumno con atributos privados y metodos públicos(get,set,__str__)
2. Define una clase Materia que pueda contener multiples objetos Alumno
3. Agrega funcionalidad para agregar alumnos a una materia
4. Implementa una clase SistemaAcademico que administre múltiples materias
5. Muestra toda la información de alumnos por materia

```python

class Alumno:
    def __init__(self, cedula, nombre, edad, nota=0):
        self.__cedula = cedula
        self.__nombre = nombre
        self.__edad = edad
        self.__nota = nota
    
    
    def get_cedula(self):
        return self.__cedula
    
    def get_nombre(self):
        return self.__nombre
    
    def get_edad(self):
        return self.__edad
    
    def get_nota(self):
        return self.__nota
    
    
    def set_nombre(self, nombre):
        self.__nombre = nombre
    
    def set_edad(self, edad):
        if edad > 0:
            self.__edad = edad
    
    def set_nota(self, nota):
        if 0 <= nota <= 5:
            self.__nota = nota
    
    
    def puede_votar(self):
        return self.__edad >= 18
    
    def obtener_calificacion(self):
        if self.__nota >= 3:
            return "Aprobado"
        else:
            return "Reprobado"
    
    
    def __str__(self):
        return f"Alumno: {self.__nombre} (Cédula: {self.__cedula}, Edad: {self.__edad}, Nota: {self.__nota})"


class Materia:
    def __init__(self, nombre, creditos, docente):
        self.nombre = nombre
        self.creditos = creditos
        self.docente = docente
        self.__alumnos = []
    
    def asignar_alumno(self, alumno):
        """Asigna un alumno a la materia"""
        if isinstance(alumno, Alumno):
            if alumno not in self.__alumnos:
                self.__alumnos.append(alumno)
                print(f"Alumno {alumno.get_nombre()} asignado a {self.nombre}")
            else:
                print(f"El alumno ya está inscrito en {self.nombre}")
        else:
            print("Error: Debe ser un objeto Alumno")
    
    def listar_estudiantes(self):
        """Lista todos los estudiantes de la materia"""
        if not self.__alumnos:
            print(f"No hay estudiantes inscritos en {self.nombre}")
            return
        
        print(f"\n=== ESTUDIANTES DE {self.nombre.upper()} ===")
        print(f"Docente: {self.docente} | Créditos: {self.creditos}")
        print("-" * 50)
        
        for i, alumno in enumerate(self.__alumnos, 1):
            estado_voto = "Puede votar" if alumno.puede_votar() else "No puede votar"
            print(f"{i}. {alumno.get_nombre()}")
            print(f"   Cédula: {alumno.get_cedula()}")
            print(f"   Edad: {alumno.get_edad()} años ({estado_voto})")
            print(f"   Nota: {alumno.get_nota()} - {alumno.obtener_calificacion()}")
            print()
    
    def get_alumnos(self):
        """Retorna la lista de alumnos"""
        return self.__alumnos
    
    def get_aprobados(self):
        """Retorna lista de alumnos aprobados"""
        return [alumno for alumno in self.__alumnos if alumno.get_nota() >= 3]
    
    def get_reprobados(self):
        """Retorna lista de alumnos reprobados"""
        return [alumno for alumno in self.__alumnos if alumno.get_nota() < 3]
    
    def __str__(self):
        return f"Materia: {self.nombre} - Docente: {self.docente} ({len(self.__alumnos)} estudiantes)"


class SistemaAcademico:
    def __init__(self):

        self.materias = []
    
    def crear_materia(self, nombre, creditos, docente):
        """Crea una nueva materia"""
        materia = Materia(nombre, creditos, docente)
        self.materias.append(materia)
        print(f"Materia '{nombre}' creada exitosamente")
        return materia
    
    def ver_materias(self):
        """Muestra todas las materias del sistema"""
        if not self.materias:
            print("No hay materias registradas")
            return
        
        print("\n=== MATERIAS EN EL SISTEMA ===")
        for i, materia in enumerate(self.materias, 1):
            print(f"{i}. {materia}")
    
    def buscar_materia(self, nombre):
        """Busca una materia por nombre"""
        for materia in self.materias:
            if materia.nombre.lower() == nombre.lower():
                return materia
        return None
    
    def resumen_general(self):
        """Muestra un resumen completo del sistema"""
        print("\n" + "="*60)
        print("           RESUMEN GENERAL DEL SISTEMA")
        print("="*60)
        
        if not self.materias:
            print("No hay materias registradas")
            return
        
        total_estudiantes = 0
        total_aprobados = 0
        total_reprobados = 0
        
        for materia in self.materias:
            estudiantes = len(materia.get_alumnos())
            aprobados = len(materia.get_aprobados())
            reprobados = len(materia.get_reprobados())
            
            total_estudiantes += estudiantes
            total_aprobados += aprobados
            total_reprobados += reprobados
            
            print(f"\n{materia.nombre}:")
            print(f"  - Docente: {materia.docente}")
            print(f"  - Estudiantes: {estudiantes}")
            print(f"  - Aprobados: {aprobados}")
            print(f"  - Reprobados: {reprobados}")
        
        print(f"\n--- TOTALES ---")
        print(f"Materias: {len(self.materias)}")
        print(f"Estudiantes: {total_estudiantes}")
        print(f"Aprobados: {total_aprobados}")
        print(f"Reprobados: {total_reprobados}")


sistema = SistemaAcademico()


matematicas = sistema.crear_materia("Matemáticas", 4, "Prof. García")
programacion = sistema.crear_materia("Programación", 3, "Prof. López")
fisica = sistema.crear_materia("Física", 4, "Prof. Martínez")

print()
sistema.ver_materias()


alumno1 = Alumno("12345678", "Ana Pérez", 19, 4.2)
alumno2 = Alumno("87654321", "Carlos Ruiz", 17, 2.8)
alumno3 = Alumno("11223344", "María González", 20, 3.7)
alumno4 = Alumno("44332211", "Luis Morales", 22, 1.5)
alumno5 = Alumno("55667788", "Sofia Herrera", 18, 4.8)

print("=== ALUMNOS CREADOS ===")
print(alumno1)
print(alumno2)
print(alumno3)
print(alumno4)
print(alumno5)


print("\n=== ASIGNANDO ALUMNOS A MATERIAS ===")


matematicas.asignar_alumno(alumno1)
matematicas.asignar_alumno(alumno2)
matematicas.asignar_alumno(alumno3)


programacion.asignar_alumno(alumno1)  
programacion.asignar_alumno(alumno4)
programacion.asignar_alumno(alumno5)


fisica.asignar_alumno(alumno2)  
fisica.asignar_alumno(alumno3)  
fisica.asignar_alumno(alumno5)  


matematicas.listar_estudiantes()
programacion.listar_estudiantes()
fisica.listar_estudiantes()


sistema.resumen_general()


print("\n=== CONSULTAS ESPECÍFICAS ===")

materia_buscada = sistema.buscar_materia("Programación")
if materia_buscada:
    print(f"Materia encontrada: {materia_buscada}")
    
    
    aprobados = materia_buscada.get_aprobados()
    reprobados = materia_buscada.get_reprobados()
    
    print(f"\nAPROBADOS en {materia_buscada.nombre}:")
    for alumno in aprobados:
        print(f"- {alumno.get_nombre()} (Nota: {alumno.get_nota()})")
    
    print(f"\nREPROBADOS en {materia_buscada.nombre}:")
    for alumno in reprobados:
        print(f"- {alumno.get_nombre()} (Nota: {alumno.get_nota()})")


print("\n=== MODIFICANDO DATOS ===")

print(f"Nota anterior de Carlos: {alumno2.get_nota()}")
alumno2.set_nota(3.2)  
print(f"Nota nueva de Carlos: {alumno2.get_nota()}")
print(f"Estado: {alumno2.obtener_calificacion()}")


print(f"\n=== VERIFICACIÓN DE VOTO ===")
alumnos_sistema = [alumno1, alumno2, alumno3, alumno4, alumno5]
for alumno in alumnos_sistema:
    voto = "SÍ" if alumno.puede_votar() else "NO"
    print(f"{alumno.get_nombre()} ({alumno.get_edad()} años): {voto} puede votar")

```
