# 🧪 Ejercicio 1: Registro de Empleados
## Enunciado:

* Crea una clase Empleado que contenga:

* Atributo público: nombre

* Atributo protegido: _salario

* Atributo privado: __id

* Un contador de empleados como atributo de clase

* Método mágico __str__ para imprimir la información del empleado

* Método estático es_valido_id que retorne True si un ID comienza con "EMP"

* Un método de clase que devuelva cuántos empleados se han creado
* También usa @property y @setter para validar que el salario no sea menor de 1,000,000.

```python

class Empleado:
    contador = 0
    
    def __init__(self, nombre, salario, id_empleado):
        self.nombre = nombre
        self._salario = salario
        self.__id = id_empleado
        
        Empleado.contador += 1
    
    
    def __str__(self):
        return f"{self.nombre}, Salario: ${self._salario}, ID: {self.__id}"
    

    @property
    def salario(self):
        return self._salario
    
    @salario.setter
    def salario(self, valor):
        if valor >= 1000000:
            self._salario = valor
        else:
            print("Salario debe ser mínimo $1,000,000")
    
    
    @staticmethod
    def es_valido_id(id_empleado):
        return id_empleado.startswith("EMP")
    

    @classmethod
    def total_empleados(cls):
        return cls.contador


emp1 = Empleado("Ana", 1500000, "EMP001")
emp2 = Empleado("Carlos", 2000000, "EMP002")


print(emp1)
print(f"Total empleados: {Empleado.total_empleados()}")

emp1.salario = 1800000  
emp1.salario = 500000   


print(f"ID válido: {Empleado.es_valido_id('EMP001')}")
print(f"ID válido: {Empleado.es_valido_id('ADM001')}")
```

# 🧪 Ejercicio 2: Cuenta Bancaria
## Enunciado:

* Crea una clase CuentaBancaria que incluya:

* Atributo público: titular

* Atributo protegido: _saldo

* Atributo privado: __numero_cuenta

* Método mágico __repr__ para representar la cuenta

* Método depositar y retirar

* @property y @setter para acceder/modificar el saldo (no permitir saldo negativo)

* Método de clase desde_string que cree una cuenta desde una cadena con formato: "Luis,1234,500000"

* Método estático es_cuenta_valida que verifique que el número de cuenta tenga 4 dígitos

```python

class CuentaBancaria:
    def __init__(self, titular, numero_cuenta, saldo_inicial=0):
        self.titular = titular
        self._saldo = saldo_inicial
        self.__numero_cuenta = numero_cuenta
    
    
    def __repr__(self):
        return f"CuentaBancaria({self.titular}, {self.__numero_cuenta}, ${self._saldo})"
    
    
    @property
    def saldo(self):
        return self._saldo
    
    @saldo.setter
    def saldo(self, valor):
        if valor >= 0:
            self._saldo = valor
        else:
            print("El saldo no puede ser negativo")
    
    
    def depositar(self, cantidad):
        if cantidad > 0:
            self._saldo += cantidad
            print(f"Depósito: ${cantidad}. Saldo actual: ${self._saldo}")
        else:
            print("La cantidad debe ser positiva")
    
    def retirar(self, cantidad):
        if cantidad > 0 and cantidad <= self._saldo:
            self._saldo -= cantidad
            print(f"Retiro: ${cantidad}. Saldo actual: ${self._saldo}")
        else:
            print("Cantidad inválida o saldo insuficiente")
    
    
    @staticmethod
    def es_cuenta_valida(numero_cuenta):
        return len(str(numero_cuenta)) == 4 and str(numero_cuenta).isdigit()
    
    
    @classmethod
    def desde_string(cls, datos):
        titular, numero, saldo = datos.split(',')
        return cls(titular, numero, int(saldo))


cuenta1 = CuentaBancaria("Ana", "1234", 100000)
print(cuenta1)

cuenta1.depositar(50000)
cuenta1.retirar(30000)
cuenta1.retirar(200000)  


cuenta2 = CuentaBancaria.desde_string("Luis,5678,500000")
print(cuenta2)


print(f"1234 es válida: {CuentaBancaria.es_cuenta_valida('1234')}")
print(f"12345 es válida: {CuentaBancaria.es_cuenta_valida('12345')}")
print(f"abc4 es válida: {CuentaBancaria.es_cuenta_valida('abc4')}")


cuenta1.saldo = -1000  
cuenta1.saldo = 200000  
print(f"Saldo actual: ${cuenta1.saldo}")

```

# 🧪 Ejercicio 3: Producto con Descuento
## Enunciado:

* Crea una clase Producto que tenga:

* Atributo público: nombre

* Atributo protegido: _precio

* Atributo privado: __codigo

* Método mágico __eq__ para comparar productos por su código

* Método estático aplicar_descuento(precio, porcentaje) que calcule el nuevo precio

* Atributo de clase impuesto que todos los productos comparten

* @property y @setter para acceder y modificar el precio (no debe ser menor de 0)

```python

class Producto:
    
    impuesto = 0.19  
    
    def __init__(self, nombre, precio, codigo):
        self.nombre = nombre
        self._precio = precio
        self.__codigo = codigo
    
    def __eq__(self, otro):
        return self.__codigo == otro.__codigo
    

    @property
    def precio(self):
        return self._precio
    
    @precio.setter
    def precio(self, valor):
        if valor >= 0:
            self._precio = valor
        else:
            print("El precio no puede ser negativo")
    
    
    @staticmethod
    def aplicar_descuento(precio, porcentaje):
        return precio * (1 - porcentaje / 100)
    
    def __str__(self):
        return f"{self.nombre}: ${self._precio} (Código: {self.__codigo})"


producto1 = Producto("Laptop", 1000000, "LAP001")
producto2 = Producto("Mouse", 50000, "MOU001")
producto3 = Producto("Laptop Dell", 1200000, "LAP001")  

print(producto1)
print(producto2)

print(f"Producto1 == Producto3: {producto1 == producto3}")  
print(f"Producto1 == Producto2: {producto1 == producto2}")  

producto1.precio = 900000
print(f"Nuevo precio: ${producto1.precio}")

producto1.precio = -1000


precio_original = 100000
precio_descuento = Producto.aplicar_descuento(precio_original, 20)
print(f"Precio original: ${precio_original}")
print(f"Con 20% descuento: ${precio_descuento}")


print(f"Impuesto general: {Producto.impuesto * 100}%")
precio_con_impuesto = producto2.precio * (1 + Producto.impuesto)
print(f"Mouse con impuesto: ${precio_con_impuesto}")

```

# 🧮 Ejercicio: Clases de Figuras Geométricas
## 🎯 Enunciado
### Crea una clase base llamada Figura con las siguientes características:

✅ Requisitos:
* Atributos:
* Público: nombre

* Protegido: _color

* Privado: __id_figura (único por figura)

### Métodos:
* Método mágico __str__ para mostrar la figura

* Método de clase crear_con_nombre que permite crear figuras por nombre (Figura.crear_con_nombre("Cuadrado"))

* Método estático es_color_valido(color) que retorna True si el color está entre ["rojo", "azul", "verde"]

* @property y @setter para acceder y modificar el color (solo si es válido)

### Luego crea una subclase Circulo que herede de Figura:

✅ Requisitos:
### Atributos:
* Protegido: _radio

### Métodos:
* Método mágico __eq__ que compara dos círculos por radio

* Método area() que devuelva el área del círculo

* @property y @setter para acceder/modificar el radio (debe ser mayor que 0)

```python

import math

class Figura:
    _contador_id = 0
    
    def __init__(self, nombre, color):
        self.nombre = nombre
        self._color = color
        Figura._contador_id += 1
        self.__id_figura = Figura._contador_id
    
    
    def __str__(self):
        return f"Figura {self.nombre} (ID: {self.__id_figura}, Color: {self._color})"
    

    @property
    def color(self):
        return self._color
    
    @color.setter
    def color(self, valor):
        if self.es_color_valido(valor):
            self._color = valor
        else:
            print("Color no válido. Debe ser: rojo, azul o verde")
    
    
    @staticmethod
    def es_color_valido(color):
        return color in ["rojo", "azul", "verde"]
    
    
    @classmethod
    def crear_con_nombre(cls, nombre):
        return cls(nombre, "rojo")  


class Circulo(Figura):
    def __init__(self, nombre, color, radio):
        super().__init__(nombre, color)
    
        self._radio = radio
    
    
    def __eq__(self, otro):
        return self._radio == otro._radio
    
    @property
    def radio(self):
        return self._radio
    
    @radio.setter
    def radio(self, valor):
        if valor > 0:
            self._radio = valor
        else:
            print("El radio debe ser mayor que 0")
    
    
    def area(self):
        return math.pi * self._radio ** 2

figura1 = Figura("Cuadrado", "azul")
print(figura1)


figura2 = Figura.crear_con_nombre("Triángulo")
print(figura2)


print(f"Rojo válido: {Figura.es_color_valido('rojo')}")
print(f"Amarillo válido: {Figura.es_color_valido('amarillo')}")


figura1.color = "verde"  # Válido
figura1.color = "amarillo"  # Inválido


circulo1 = Circulo("Círculo A", "rojo", 5)
circulo2 = Circulo("Círculo B", "azul", 5)
circulo3 = Circulo("Círculo C", "verde", 3)

print(circulo1)
print(f"Área: {circulo1.area():.2f}")


print(f"Círculo1 == Círculo2: {circulo1 == circulo2}")  # True (mismo radio)
print(f"Círculo1 == Círculo3: {circulo1 == circulo3}")  # False

circulo1.radio = 10  # Válido
print(f"Nuevo radio: {circulo1.radio}")
print(f"Nueva área: {circulo1.area():.2f}")

circulo1.radio = -5  # Inválido

```
